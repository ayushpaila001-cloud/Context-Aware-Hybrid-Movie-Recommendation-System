# Netflix Hybrid Recommendation System

## Overview

This project presents a **Hybrid Movie Recommendation System** that combines **Content-Based Filtering** and **User-Based Collaborative Filtering** to generate personalized movie recommendations.

Unlike traditional recommendation systems that rely solely on viewing history or movie metadata, this approach integrates both user preferences and content characteristics to improve recommendation quality.

The project demonstrates the complete recommendation pipeline, including:

- Data preprocessing
- Exploratory Data Analysis (EDA)
- User behavior analysis
- Content vector creation
- User vector creation
- Cosine similarity computation
- Hybrid recommendation scoring
- Personalized movie recommendations

The implementation is developed entirely in **Python** using **Pandas**, **NumPy**, and **Scikit-learn**.

# User Decision Fatigue Score

One of the objectives of this project is to quantify the level of decision fatigue experienced while selecting content.

A custom fatigue score is developed using multiple behavioral indicators, including:

- Decision time
- Number of titles browsed
- Choice overload
- Clarity before watching
- Skipping long content

The resulting score classifies users into different fatigue levels, enabling more personalized recommendations.

```python
decision_time_map = {
    'Less than 5 minutes': 1,
    '5-10 minutes': 2,
    '10-20 minutes': 3,
    '20-30 minutes': 4,
    'More than 30 minutes': 5
}

df['decision_time_num'] = df['decision_time'].map(decision_time_map)
```

```python
titles_map = {
    '1-5': 1,
    '6-10': 2,
    '11-20': 3,
    'More than 20': 4
}

df['titles_browsed_num'] = df['titles_browsed'].map(titles_map)
```

```python
overload_map = {
    'Never': 1,
    'Rarely': 2,
    'Sometimes': 3,
    'Often': 4,
    'Almost Always': 5
}

df['choice_overload_num'] = df['choice_overload'].map(overload_map)
```

```python
skip_map = {
    'Never':1,
    'Rarely':2,
    'Sometimes':3,
    'Frequently':4
}

df['skip_long_content_num'] = df['skip_long_content'].map(skip_map)
```

```python
df['clarity_inverted'] = 6 - df['clarity_before_watching']
```

```python
df['fatigue_score'] = (
    df['decision_time_num']*0.30 +
    df['titles_browsed_num']*0.25 +
    df['clarity_inverted']*0.20 +
    df['choice_overload_num']*0.15 +
    df['skip_long_content_num']*0.10
)
```

```python
min_score = df['fatigue_score'].min()
max_score = df['fatigue_score'].max()

df['fatigue_score'] = (
    (df['fatigue_score'] - min_score) /
    (max_score - min_score)
) * 100
```

```python
def fatigue_level(score):

    if score < 30:
        return 'Low'

    elif score < 55:
        return 'Moderate'

    elif score < 75:
        return 'High'

    else:
        return 'Severe'

df['fatigue_tier'] = df['fatigue_score'].apply(fatigue_level)
```

#  Statistical Validation

To determine whether observed behavioral patterns are statistically significant, Chi-Square tests of independence are performed.

These tests evaluate whether relationships between categorical variables occur by chance or represent meaningful associations within the dataset.

```python
from scipy.stats import chi2_contingency
```

```python
table = pd.crosstab(
    df['fatigue_tier'],
    df['exit_without_watching']
)

table
```

```python
chi2, p, dof, expected = chi2_contingency(table)
```

```python
print("Chi-Square Statistic =", chi2)

print("P-value =", p)

print("Degrees of Freedom =", dof)

print("Expected Frequencies")

print(expected)
```

```python
table = pd.crosstab(
    df['fatigue_tier'],
    df['recommendation_match']
)

table
```

```python
print("Chi-Square :", round(chi2,2))
print("P-value :", round(p,4))
print("Degrees of Freedom :", dof)
```

# User Preference Vector

A user vector is created to represent individual viewing preferences.This representation allows users to be directly compared with movies in the feature space.

```python
user_vectors_df.shape
```

```python
for i, row in user_df.iterrows():

    feature = f"type_{row['Preferred_Format']}"

    if feature in user_vectors_df.columns:
        user_vectors_df.loc[i, feature] = 1
```

```python
user_vectors_df.head()
```

```python
for i, row in user_df.iterrows():

    languages = row["Languages"].split(",")

    for lang in languages:

        feature = f"language_{lang.strip()}"

        if feature in user_vectors_df.columns:
            user_vectors_df.loc[i, feature] = 1
```

#  Hybrid Recommendation Model

A weighted hybrid recommendation strategy combines multiple scoring components to improve recommendation quality.Each feature contributes according to its assigned weight, producing a personalized ranking for every candidate movie.

```python
WEIGHTS = {

    "similarity": 0.60,

    "mood": 0.15,

    "duration": 0.10,

    "watching": 0.07,

    "day": 0.04,

    "language": 0.03,

    "time": 0.01
}
```

```python
duration_mapping = {
    "Short": 45,
    "Medium": 90,
    "Long": 150
}

def get_duration_score(movie, session):

    user_time = session["available_time_minutes"]

    movie_time = duration_mapping[movie["duration_category"]]

    difference = abs(user_time - movie_time)

    if difference <= 15:
        return 1.0

    elif difference <= 45:
        return 0.7

    elif difference <= 75:
        return 0.4

    else:
        return 0.0
```

```python
def get_language_score(movie, session):

    if movie["language"] == session["preferred_language_today"]:
        return 1.0

    return 0.0
```

```python
watching_rules = {

    "Alone": [
        "Thriller",
        "Mystery",
        "Mind-Bending",
        "Dark",
        "Sci-Fi"
    ],

    "Partner": [
        "Romantic",
        "Feel-Good",
        "Comedy"
    ],

    "Friends": [
        "Action",
        "Comedy",
        "Mass",
        "High-Octane"
    ],

    "Family": [
        "Family",
        "Comedy",
        "Adventure",
        "Feel-Good"
    ]
}
```

```python
def get_watching_score(movie, session):

    group = session["watching_with"]

    genres = [
        movie["genre"],
        movie["mood_1"],
        movie["mood_2"]
    ]

    if group not in watching_rules:
        return 0

    for g in genres:

        if g in watching_rules[group]:
            return 1

    return 0
    
```

```python
def get_day_score(movie, session):

    if session["day_type"] == "Weekend":

        if movie["duration_category"] == "Long":
            return 1

        return 0.7

    else:

        if movie["duration_category"] == "Short":
            return 1

        if movie["duration_category"] == "Medium":
            return 0.7

        return 0.4
```

```python
time_rules = {

    "Morning": [
        "Comedy",
        "Family",
        "Inspirational"
    ],

    "Afternoon": [
        "Adventure",
        "Comedy",
        "Action"
    ],

    "Evening": [
        "Thriller",
        "Action",
        "Drama",
        "Romantic"
    ],

    "Night": [
        "Dark",
        "Mystery",
        "Mind-Bending",
        "Thriller"
    ]
}
```

```python
def get_time_score(movie, session):

    time = session["time_of_day"]

    genres = [
        movie["genre"],
        movie["mood_1"],
        movie["mood_2"]
    ]

    if time not in time_rules:
        return 0

    for g in genres:

        if g in time_rules[time]:
            return 1

    return 0
```

```python
def calculate_final_score(movie_name, user_id):

    movie = get_movie_details(movie_name)

    session = get_current_session(user_id)

    similarity = similarity_df.loc[user_id, movie_name]
    mood_score = get_mood_score(movie, session)

    duration_score = get_duration_score(movie, session)

    watching_score = get_watching_score(movie, session)

    day_score = get_day_score(movie, session)

    language_score = get_language_score(movie, session)

    time_score = get_time_score(movie, session)

    # Final Weighted Scor
    final_score = (

        similarity * WEIGHTS["similarity"]

        + mood_score * WEIGHTS["mood"]

        + duration_score * WEIGHTS["duration"]

        + watching_score * WEIGHTS["watching"]

        + day_score * WEIGHTS["day"]

        + language_score * WEIGHTS["language"]

        + time_score * WEIGHTS["time"]

    )

    return final_score
    
```

```python
calculate_final_score("RRR", "U001")
```

```python
def rerank_movies(user_id):

    candidates = get_candidate_movies(user_id)

    scores = {}

    for movie_name in candidates.index:

        score = calculate_final_score(movie_name, user_id)

        scores[movie_name] = score

    reranked = pd.Series(scores)

    return reranked.sort_values(ascending=False)
```

```python
def recommend_movies(user_id, top_n=10):

    recommendations = rerank_movies(user_id)

    return recommendations.head(top_n).index.tolist()
```

# Generating Personalized Recommendations

* The final step ranks all candidate movies based on their hybrid recommendation score.

* Movies already watched by the user are excluded, ensuring that only unseen and relevant recommendations are returned.

* The resulting ranked list represents the final personalized recommendation output.

```python
def recommend_movies_table(user_id, top_n=10):

    movies = rerank_movies(user_id).head(top_n)

    result = pd.DataFrame({
        "Rank": range(1, top_n + 1),
        "Recommended Movie": movies.index
    })

    return result
```
